- **Collaborative Features**

  Share notebooks via email, GitHub, or other collaboration platforms. Convert notebooks to various formats (HTML, PDF, slideshows) for presentation and sharing.

### **2.1 Basic `for` loop **

A basic `for` loop iterates over a sequence (such as a list, tuple, or string) and executes a block of code for each item in the sequence.

*Example 1: iterating over a list*

```python
# List of fruits
fruits = ["apple", "banana", "cherry"]

# For loop to iterate over the list
for fruit in fruits:
    print(fruit)
```

*Example 2: Iterating over a string*

```python
# String
word = "hello"

# For loop to iterate over each character in the string
for char in word:
    print(char)
```

*Example 3: Interating over a range of numbers*

```python
# For loop to iterate over a range of numbers
for i in range(5):
    print(i)
```

*Example 4: Nested for loops*

```python
# Nested for loops to print a multiplication table
for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end="\t")
    print()
```

## **Optional Task: Write a Python Function to Transpose a Matrix**


Define a function named `transpose_matrix`. The function should take a matrix (a list of lists) as input.

```python
# Input matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Output matrix (transpose)
# [
#     [1, 4, 7],
#     [2, 5, 8],
#     [3, 6, 9]
# ]
```

```python
def transpose_matrix(matrix):
    width = len(matrix)
    height = len(matrix[0])
    result = []
    result_width = []
    for i in range(width):
        for j in range(height):
            result_width.append(matrix[j][i])
        result.append(result_width)
        result_width = []
    return result

# test

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

transposed_matrix = transpose_matrix(matrix)

# Print the original matrix
print("Original Matrix:")
for row in matrix:
    print(row)

# Print the transposed matrix
print("\nTransposed Matrix:")
for row in transposed_matrix:
    print(row)
```

```python
# insert your code here

def transpose_matrix(matrix):
    """
    This function takes a matrix (a list of lists) and returns its transpose.
    """
    # Get the number of rows and columns
    rows = len(matrix)
    cols = len(matrix[0])

    # Initialize an empty matrix for the transpose
    transpose = []

    # Use nested for loops to create the transposed matrix
    for col in range(cols):
        # Create a new row for the transposed matrix
        new_row = []
        for row in range(rows):
            # Append the element from the original matrix to the new row
            new_row.append(matrix[row][col])
        # Append the new row to the transposed matrix
        transpose.append(new_row)

    return transpose

# Test the function with the example matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transposed_matrix = transpose_matrix(matrix)

# Print the original matrix
print("Original Matrix:")
for row in matrix:
    print(row)

# Print the transposed matrix
print("\nTransposed Matrix:")
for row in transposed_matrix:
    print(row)
```

### **Optional Reference: Timing It**

The `%%timeit` cell magic repeats a code block and reports timing statistics. If `-n` is not provided, IPython chooses a loop count automatically. Here we use small fixed values so the classroom example finishes predictably.


```python
ufo5 = ufo.copy()
ufo5["City"] = ufo5["City"].fillna("No city")
ufo5
```

```python
ufo5['place']=''
ufo5['address']=''
ufo5.head()
```

```python
%%timeit -n 3 -r 3

ufo5['place'] = ufo5['City'] + ', ' + ufo5['State']
```

```python
%%timeit -n 3 -r 3

for i in ufo5.index:
    ufo5.iloc[i,6] = ufo5.iloc[i,0] + ', ' + ufo5.iloc[i,3]
```

Timing results can vary between runs, but the row-by-row loop should be substantially slower than the vectorised Pandas operation.

**Note**: Pandas is based on NumPy arrays, so prefer vectorised operations when a clear one is available.
