## **Week 4 Applied Session: Regular Expression**

In the lecture, you have learnt how to use regular expressions. To quickly review some regular expression syntax:


* <font color="red">[0-9]</font> Matches a single digit
* <font color="red">[a-z0-9]</font> Matches a single character that must be a lower case letter or a digit.
* <font color="red">[A-Za-z]</font> Matches a single character that must be an upper/lower case letter
* <font color="red">\d</font> Matches any decimal digit; equivalent to the set [0-9].
* <font color="red">\D</font> Matches characters that are not digits, which is equivalent to [^0-9] or [^\d].
* <font color="red">\w</font> Matches any alphanumeric character, which is equivalent to [a-zA-Z0-9].
* <font color="red">\W</font> Matches any non-alphanumeric character; which is equivalent to [^a-zA-Z0-9] or [^\w].
* <font color="red">\s</font> Matches any whitespace character; which is equivalent to [\t\n\r\f\v], where \t indicates tabs, \n  line feeds, \r carriage returns, \f form feeds and \v vertical tabs.
* <font color="red">\S:</font> Matches any non-whitespace character; which is equivalent to  [^ \t\n\r\f\v].
* <font color="red">ˆ</font> Matches the start of the line.
* <font color="red">$</font> Matches the end of the line.
* <font color="red">.</font> Matches any character (a wildcard).
* <font color="red">*</font> Matches when the preceding character occurs zero or more times
* <font color="red">?</font> Matches when the preceding character occurs zero or one times
* <font color="red">+</font> Matches when the preceding character occurs one or more times

More information can be found here :

### **1. Backslash**

**What is `\`?**

`\`, backslash or escape-character, is used to indicate special forms or to allow special characters to be used without invoking their special meaning.



**How about `r""` ? When to use it?**

`r""` (or `r''`) is Python's raw-string prefix. It affects how Python parses the string literal, not how the regex engine works. Using a raw string keeps backslashes as literal characters so you don't need to double escape.

For example, `"\\t"` represents the two characters `\` and `t`, whereas `"\t"` represents a tab character.

Sometimes you can use them interchangeably,

````python
str1 = re.findall('\\t', "Please find \t")
print (str1)

str2 = re.findall(r'\\t', "Please find \\t")
print (str2)
````

Refine the regular expression for date to distinguish months with 29/30/31 days.

**Note**: assume all years are a leap year, which means every February has 29 days.

### **Understanding Verbose Regular Expressions: `(?x)` flag**

When regular expressions become complex, they can be hard to read and maintain. Python provides the **verbose mode** (or extended mode) to make them more readable.

**Two ways to enable verbose mode:**
1. Inline flag: `(?x)` at the start of the pattern
2. Flag parameter: `re.VERBOSE` or `re.X` as a flag

**What verbose mode does:**
- **Ignores whitespace** (spaces, tabs, newlines) in the pattern
- **Allows comments** using `#` (everything after `#` to end of line is ignored)
- Makes complex patterns much more readable

**Example:**

Without verbose mode:
```python
r'([12][0-9]|0?[1-9])[/-](0?2)|(30|[12][0-9]|0?[1-9])[/-](0?[469]|11)'
```

With verbose mode:
```python
r'''(?x)
    (?:
    # February (29 days)
      ([12][0-9]|0?[1-9])[/-](0?2)
    # 30-day months
      |(30|[12][0-9]|0?[1-9])[/-](0?[469]|11)
    )
'''
```

**Note:** If you need to match literal whitespace in verbose mode, use `\ ` (backslash-space) or `\s`.

````python
# define a function to validate the input date
def date(pattern, m):
    if re.match(pattern, m):
        print (m + " is a date")
    else:
        print (m + " is NOT a date")
````

````python
regex = r'''(?x)
    (?:
    # February (29 days every year)
      ([12][0-9]|0?[1-9])[/-](0?2)
    # 30-day months
      |(30|[12][0-9]|0?[1-9])[/-](0?[469]|11)
    # 31-day months
      |(3[01]|[12][0-9]|0?[1-9])[/-](0?[13578]|1[02])
    )
    # Year
    [/-]((?:[0-9]{2})?[0-9]{2})
'''
````

````python
#date(r"((31[/-](0?[13578]|1[02]))|(30[/-](0?[469]|11))|(28[/-]02))[/-]((?:\d{2})?\d{2})", "28/02/2019")
date(regex, "28/02/2019")
date(regex, "31/04/2019")
date(regex, "29/05/2019")
date(regex, "31/06/2019")
````

### Practice 1: Anchors, flags, and groups

Quick refresher on anchors, flags, and grouping.

````python
# Anchors ^ and $; multiline flag
text = """line1
Start: value
line3
start: other
END"""

print(re.findall(r"^start:", text))  # no match (case sensitive, single-line)
print(re.findall(r"^start:", text, flags=re.I|re.M))  # multiline + ignorecase

# Capturing vs non-capturing groups; alternation
m = re.findall(r"(cat|dog)s?", "cats, dogs, dog")
print(m)

m2 = re.findall(r"(?:cat|dog)s?", "cats, dogs, dog")  # non-capturing
print(m2)

# Backreferences
print(bool(re.search(r"(\w+) \1", "go go gadget")))  # True
print(bool(re.search(r"(\w+) \1", "go gadget")))      # False
````

### Practice 2: Lookarounds (no consumption)

Use lookaheads and lookbehinds to assert context without including it in the match.


````python
text = "Order#12345 paid; Order#67890 pending; Ref#555"

# Positive lookbehind to grab numbers after 'Order#'
print(re.findall(r"(?<=Order#)\d+", text))  # ['12345', '67890']

# Negative lookahead: words not followed by ':'
text2 = "key: value; token abc; id: 42"
print(re.findall(r"\b\w+\b(?!:)", text2))  # ['value', 'token', 'abc', '42']

# Extract emails only from lines where Status=Active using lookahead across line
records = """user=a@x.com Status=Active
user=b@y.com Status=Inactive
user=c@z.com Status=Active"""
print(re.findall(r"(?m)^user=([\w.+-]+@[\w.-]+)\b(?=.*Status=Active)$", records))
````

### Practice 3: Named groups and compiled patterns

Improve readability and performance when reusing patterns.


````python
log_line = "2025-08-12 14:33:01,123 INFO module=user action=login user_id=42"

pattern = re.compile(r"""
    (?P<ts>\d{4}-\d{2}-\d{2}\s\d{2}:\d{2}:\d{2},\d{3})\s+
    (?P<level>INFO|WARN|ERROR)\s+
    (?P<context>[^\s]+)\s+
    module=(?P<module>\w+)\s+
    action=(?P<action>\w+)\s+
    user_id=(?P<user_id>\d+)
""", re.X)

m = pattern.search(log_line)
if m:
    print(m.groupdict())

# Reuse the same pattern on multiple lines
logs = [
    "2025-08-12 14:33:01,123 INFO module=user action=login user_id=42",
    "2025-08-12 14:35:10,777 ERROR module=auth action=deny user_id=42",
]
print([pattern.search(x).group("level") for x in logs])
````

### Practice 4: Stricter IPv4 and common pitfalls

Make IPv4 more precise and show why simple patterns overmatch.


````python
# Basic IPv4 matches 999.999.999.999 (overmatch)
basic_ip = re.compile(r"\b(?:\d{1,3}\.){3}\d{1,3}\b")

# Strict IPv4 restricts each octet to 0-255
strict_ip = re.compile(r"""
    \b(
        (25[0-5])        # 250-255
      | (2[0-4]\d)      # 200-249
      | (1?\d?\d)      # 0-199
    )(\.(
        (25[0-5])|(2[0-4]\d)|(1?\d?\d)
    )){3}\b
""", re.X)

samples = [
    "valid 0.0.0.0 and 192.168.1.1",
    "invalid 256.1.1.1 and 999.10.10.10",
]
for s in samples:
    print("basic:", re.findall(basic_ip, s))
    print("strict:", [m[0] for m in re.findall(strict_ip, s)])
````

### Practice 5: Using pandas `.str` with regex

Vectorised extraction, filtering, and replacement.


````python
data = pd.DataFrame({
    "text": [
        "Order O-1001: paid by alice@example.com",
        "Order O-1002: pending by bob.smith@corp.co.uk",
        "Ref R-9: unknown",
    ]
})

# Extract order ids and emails
out = data["text"].str.extract(r"Order\s+(?P<oid>[A-Z]-\d+):.*?(?P<email>[\w.+-]+@[\w.-]+)", expand=True)
print(out)

# Filter rows with emails and replace domain
mask = data["text"].str.contains(r"@[\w.-]+", na=False)
print(data[mask])

print(data["text"].str.replace(r"@.+$", "@redacted", regex=True))
````

### Practice 7: Greedy vs lazy quantifiers

Demonstrate `.*` vs `.*?` and when to use each.


````python
html = "<div><p>one</p><p>two</p></div>"
print(re.findall(r"<p>.*</p>", html))     # greedy, one big match
print(re.findall(r"<p>.*?</p>", html))    # lazy, two matches

# Use DOTALL if needed across lines
html2 = "<p>line1\nline2</p>"
print(re.findall(r"<p>.*?</p>", html2))             # no match without DOTALL
print(re.findall(r"<p>[\s\S]*?</p>", html2))      # works without flags
print(re.findall(r"<p>.*?</p>", html2, flags=re.S)) # DOTALL
````
